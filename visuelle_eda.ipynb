## Data quality and completeness

Checks on the VISUELLE sales, price/discount, restock, customer and weather tables: missing values and duplicates, key integrity between tables, value validity, date coverage and gaps, and category sample sizes. Run after the loading cells above.

```python
import numpy as np
import pandas as pd

# Data quality and completeness checks across all loaded tables.
tables = {
    "sales": sales_df, "price_discount": pds_df, "customer": cus_df,
    "restocks": restocks_df, "weather": vis_weather_df, "search" : vis_gtrends_df
}

# 1. Completeness and duplicates per table.
overview = pd.DataFrame({
    name: {
        "rows": len(df),
        "columns": df.shape[1],
        "cells_missing": int(df.isna().sum().sum()),
        "columns_with_missing": int((df.isna().sum() > 0).sum()),
        "missing_%": round(100 * df.isna().sum().sum() / df.size, 4),
        "duplicate_rows": int(df.duplicated().sum()),
    }
    for name, df in tables.items()
}).T
print("1. Completeness and duplicates")
display(overview)

missing_cols = {
    name: df.isna().sum()[lambda s: s > 0].to_dict() for name, df in tables.items()
}
print("Columns with missing values:", {k: v for k, v in missing_cols.items() if v})

# 2. Key integrity between tables.
keys = ["external_code", "retail"]
sales_keys = sales_df[keys].drop_duplicates()
pds_keys = pds_df[keys].drop_duplicates()
restock_keys = restocks_df[keys].drop_duplicates()
integrity = pd.Series({
    "sales: duplicate (product, retailer) keys": int(sales_df.duplicated(keys).sum()),
    "price_discount: duplicate keys": int(pds_df.duplicated(keys).sum()),
    "sales keys missing from price_discount": len(sales_keys.merge(pds_keys, on=keys, how="left", indicator=True).query("_merge == 'left_only'")),
    "price_discount keys missing from sales": len(pds_keys.merge(sales_keys, on=keys, how="left", indicator=True).query("_merge == 'left_only'")),
    "restock keys missing from sales": len(restock_keys.merge(sales_keys, on=keys, how="left", indicator=True).query("_merge == 'left_only'")),
    "sales pairs with no restock records": len(sales_keys.merge(restock_keys, on=keys, how="left", indicator=True).query("_merge == 'left_only'")),
    "customer products not in sales catalog": int((~cus_df["external_code"].isin(sales_df["external_code"])).sum()),
    "customer retailers not in sales": int((~cus_df["retail"].isin(sales_df["retail"])).sum()),
    "products without a unique category": int((sales_df.groupby("external_code")["category"].nunique() > 1).sum()),
})
print("\n2. Key integrity")
display(integrity.rename("count").to_frame())

# 3. Value validity.
sales_months = [str(m) for m in range(12)]
validity = pd.Series({
    "sales: negative monthly units": int((sales_df[sales_months] < 0).sum().sum()),
    "sales: non-integer monthly units": int((sales_df[sales_months] % 1 != 0).sum().sum()),
    "sales: all-zero product-retailer rows": int((sales_df[sales_months].sum(axis=1) == 0).sum()),
    "discount: outside 0-1": int(((pds_df[sales_months] < 0) | (pds_df[sales_months] > 1)).sum().sum()),
    "price: not positive": int((pds_df["price"] <= 0).sum()),
    "restocks: non-positive qty": int((restocks_df["qty"] <= 0).sum()),
    "restocks: invalid ISO week (not 1-53)": int((~restocks_df["week"].between(1, 53)).sum()),
    "customer: non-positive qty": int((cus_df["qty"] <= 0).sum()),
    "weather: humidity outside 0-100": int(((vis_weather_df["humidity %"] < 0) | (vis_weather_df["humidity %"] > 100)).sum()),
    "weather: min temp > max temp": int((vis_weather_df["min temp °C"] > vis_weather_df["max temp °C"]).sum()),
})
print("\n3. Value validity")
display(validity.rename("count").to_frame())

# 4. Date coverage and gaps.
cust_dates = pd.to_datetime(cus_df["data"], errors="coerce")
weather_dates = pd.to_datetime(vis_weather_df["date"], format="%d/%m/%Y", errors="coerce")
release_dates = pd.to_datetime(sales_df["release_date"], errors="coerce")
restock_dates = pd.to_datetime(
    restocks_df["year"].astype(str) + "-" + restocks_df["week"].astype(str) + "-1",
    format="%G-%V-%u", errors="coerce",
)
coverage = pd.DataFrame({
    "start": [release_dates.min(), restock_dates.min(), cust_dates.min(), weather_dates.min()],
    "end": [release_dates.max(), restock_dates.max(), cust_dates.max(), weather_dates.max()],
    "unparseable_dates": [release_dates.isna().sum(), restock_dates.isna().sum(), cust_dates.isna().sum(), weather_dates.isna().sum()],
}, index=["sales release_date", "restocks", "customer transactions", "weather (daily)"])
print("\n4. Date coverage")
display(coverage)

monthly_cust = cust_dates.dt.to_period("M").value_counts().sort_index()
full_months = pd.period_range(monthly_cust.index.min(), monthly_cust.index.max(), freq="M")
print("Months with no customer transactions:", [str(m) for m in full_months.difference(monthly_cust.index)])
print("Months with under 10% of the median monthly transactions:", [str(m) for m in monthly_cust[monthly_cust < 0.1 * monthly_cust.median()].index])

# 5. Sample size per category.
category_rows = sales_df.groupby("category").size().sort_values()
print("\n5. Categories with fewer than 1,000 product-retailer rows:", category_rows[category_rows < 1000].to_dict())

# 6. Summary of issues that matter for modelling.
flags = []
if overview["duplicate_rows"].sum():
    flags.append(f"{int(overview['duplicate_rows'].sum())} duplicate rows across tables")
if validity["sales: negative monthly units"]:
    flags.append(f"{validity['sales: negative monthly units']} negative monthly sales values (returns?)")
if integrity["sales pairs with no restock records"]:
    flags.append(f"{integrity['sales pairs with no restock records']} product-retailer pairs with no restocks")
if integrity["customer products not in sales catalog"]:
    flags.append(f"{integrity['customer products not in sales catalog']} customer rows for products not in the catalog")
print("\nFlags:")
for f in flags:
    print(" -", f)
```

**Summary: data quality and completeness**
- **Complete overall:** sales, restocks and customer tables have no missing cells. Weather has 3.2% missing (3,368 days with no readings) and price_discount has one missing value.
- **Keys are consistent between sales and price/discount**; only 11 sales pairs have no restock records.
- **Orphan records:** restocks include 272,723 product-retailer pairs not in the sales table, and 44,614 customer rows reference products not in the catalog. Join on the sales catalog before modelling.
- **Value issues:** 3,386 negative monthly sales values (likely returns) and one duplicate customer row. No invalid discounts, prices, quantities or weeks.
- **Coverage gaps:** customer transactions have no records in 2020-04 and 2020-11 and very low volume from 2020-03 onward, so models should not rely on that period.
- **Small categories:** nine categories have under 1,000 product-retailer rows, so their trend estimates are noisy.

## EDA



The first plot shows monthly unit sales for every styling category in separate panels, ordered by total sales. The second summarizes average product-retailer monthly sales across discount bands.


```python
import matplotlib.pyplot as plt

sales_df = pd.read_csv("data/visuelle2/sales.csv")
restocks_df = pd.read_csv("data/visuelle2/restocks.csv")

keys = ["external_code", "retail"]
sales_months = [str(i) for i in range(12)]

sales_df["sales_total"] = sales_df[sales_months].sum(axis=1)

restocks_by_item = (
    restocks_df.groupby(keys, as_index=False)
    .agg(restock_total=("qty", "sum"))
)

comparison = sales_df.merge(
    restocks_by_item,
    on=keys,
    how="inner",
    validate="one_to_one",
)

print("Matched product-retailer pairs:", len(comparison))
print("Correlation:", comparison["sales_total"].corr(comparison["restock_total"]))

comparison.plot.scatter(x="sales_total", y="restock_total", alpha=0.2)
plt.title("Total sales vs. total restocks")
plt.show()
```

**Summary: total sales vs. total restocks**
- Across 106,839 matched product-store pairs, total sales and total restocks are strongly correlated (r = 0.92).
- Stores restock what sells, so restock volume is a useful stock signal and a proxy for demand.
- The correlation is partly mechanical: the same demand drives both, so it should not be read as restocks causing sales.

Sales over the 12 monthly columns, by styling category — plot total monthly unit sales (columns 0–11) in a separate panel for each category. Independent y-axes make trends legible across categories with different sales volumes.


Discount level vs. sales — reshape the monthly columns in sales.csv and price_discount_series.csv into one row per product-month, then plot average sales by discount band. This is more informative than comparing annual totals.

```python
import matplotlib.pyplot as plt
import pandas as pd

sales_months = [str(month) for month in range(12)]

# Plot a separate monthly sales trend for every styling category.
category_monthly_sales = sales_df.groupby("category", observed=True)[sales_months].sum()
category_order = category_monthly_sales.sum(axis=1).sort_values(ascending=False).index
category_monthly_sales = category_monthly_sales.loc[category_order]

ncols = 4
nrows = (len(category_monthly_sales) + ncols - 1) // ncols
fig, axes = plt.subplots(nrows, ncols, figsize=(16, 3 * nrows), sharex=True, squeeze=False)
for ax, (category, monthly_sales_by_category) in zip(axes.flat, category_monthly_sales.iterrows()):
    ax.plot(range(1, 13), monthly_sales_by_category, marker="o", linewidth=1.5)
    ax.set_title(f"{category} ({monthly_sales_by_category.sum():,.0f} units)", fontsize=9)
    ax.set_xticks(range(1, 13))
    ax.grid(axis="y", alpha=0.25)
    ax.tick_params(axis="x", labelsize=8)

for ax in axes.flat[len(category_monthly_sales):]:
    ax.set_visible(False)
fig.supxlabel("Month")
fig.supylabel("Units sold")
fig.suptitle("Monthly sales trend for every styling category", y=1.01)
fig.tight_layout()
plt.show()

# Compare product-retailer monthly sales with their corresponding discount.
monthly_sales = sales_df.melt(
    id_vars=["external_code", "retail"],
    value_vars=sales_months,
    var_name="month",
    value_name="units_sold",
)
monthly_discounts = pds_df.melt(
    id_vars=["external_code", "retail"],
    value_vars=sales_months,
    var_name="month",
    value_name="discount",
)
sales_discount_monthly = monthly_sales.merge(
    monthly_discounts,
    on=["external_code", "retail", "month"],
    how="inner",
    validate="one_to_one",
)

discount_bands = [-0.001, 0, 0.1, 0.2, 0.3, 0.5, 1.0]
discount_labels = ["0%", "0-10%", "10-20%", "20-30%", "30-50%", "50-100%"]
sales_discount_monthly["discount_band"] = pd.cut(
    sales_discount_monthly["discount"],
    bins=discount_bands,
    labels=discount_labels,
)
discount_summary = (
    sales_discount_monthly.dropna(subset=["discount_band"])
    .groupby("discount_band", observed=False)["units_sold"]
    .agg(["mean", "median", "count"])
)
ax = discount_summary["mean"].plot(kind="bar", figsize=(9, 5), color="steelblue")
ax.set_ylabel("Average units sold per product-retailer-month")
ax.set_xlabel("Discount")
ax.set_title("Average monthly sales by discount band")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

discount_summary
```

**Summary: monthly sales by styling category and by discount band**
- Each of the 27 categories has its own monthly sales curve, and volumes differ widely (culottes, long sleeve, doll dress and long dress lead), so categories need separate views.
- Average units per product-store-month peak at 0-10% discount (about 2.1) and fall to about 1.1 at 30%+.
- Deep discounts come late in a product's life, when demand is already fading, so this reflects timing rather than discounts reducing sales.

```python
# Average monthly units by styling category and discount band.
category_lookup = sales_df[["external_code", "category"]].drop_duplicates("external_code")
category_discount = sales_discount_monthly.merge(
    category_lookup, on="external_code", how="left", validate="many_to_one"
).dropna(subset=["discount_band", "category"])

category_band_mean = category_discount.pivot_table(
    index="category", columns="discount_band", values="units_sold", aggfunc="mean", observed=False
)
category_band_rows = category_discount.pivot_table(
    index="category", columns="discount_band", values="units_sold", aggfunc="size", observed=False
).fillna(0)
category_order = category_discount.groupby("category")["units_sold"].sum().sort_values(ascending=False).index
category_band_mean = category_band_mean.loc[category_order]
# Hide cells with too few observations to be meaningful.
category_band_mean = category_band_mean.where(category_band_rows.loc[category_order] >= 30)

fig, ax = plt.subplots(figsize=(9, 10))
im = ax.imshow(category_band_mean.to_numpy(dtype=float), cmap="YlGnBu", aspect="auto")
ax.set_xticks(range(len(category_band_mean.columns)), category_band_mean.columns)
ax.set_yticks(range(len(category_band_mean.index)), category_band_mean.index)
for i in range(category_band_mean.shape[0]):
    for j in range(category_band_mean.shape[1]):
        value = category_band_mean.iat[i, j]
        if pd.notna(value):
            ax.text(j, i, f"{value:.1f}", ha="center", va="center", fontsize=8)
ax.set_xlabel("Discount band")
ax.set_title("Average units sold per product-store-month\nby styling category and discount band (blank = fewer than 30 rows)")
fig.colorbar(im, ax=ax, label="Average units")
fig.tight_layout()
plt.show()

category_band_mean.round(2)
```

**Summary: average units by category and discount band**
- Every category follows the same pattern as the overall chart: sales peak in the 0-10% band and decline as discounts deepen.
- Level differs by category: for example long dress reaches about 2.9 units at 0-10%, while long sleeve stays near 1.0-1.7.
- This supports using discount as a model feature, ideally interacting with category, but it is descriptive, not causal. Blank cells have fewer than 30 observations.

```python
# Underlying category momentum: sales relative to what an average product sells at the same age and discount band.
momentum_df = sales_discount_monthly.dropna(subset=["discount_band"]).merge(
    sales_df[["external_code", "retail", "category", "season"]],
    on=["external_code", "retail"], how="left", validate="many_to_one",
)
momentum_df["release_year"] = 2000 + momentum_df["season"].str[-2:].astype(int)

# Expected units = average for the same months-since-release and discount band across all categories.
momentum_df["expected_units"] = momentum_df.groupby(["month", "discount_band"], observed=True)["units_sold"].transform("mean")

latest_year = momentum_df["release_year"].max()
period = momentum_df["release_year"].map({latest_year: "latest", latest_year - 1: "previous"})
momentum_df = momentum_df.assign(period=period).dropna(subset=["period"])

by_period = momentum_df.groupby(["category", "period"], observed=True).agg(
    actual=("units_sold", "sum"), expected=("expected_units", "sum"), rows=("units_sold", "size"),
).unstack("period")
category_momentum = pd.DataFrame({
    "raw_growth_pct": 100 * (by_period[("actual", "latest")] / by_period[("actual", "previous")] - 1),
    "index_previous": by_period[("actual", "previous")] / by_period[("expected", "previous")],
    "index_latest": by_period[("actual", "latest")] / by_period[("expected", "latest")],
    "min_rows": by_period["rows"].min(axis=1),
})
category_momentum["adjusted_momentum_pct"] = 100 * (category_momentum["index_latest"] / category_momentum["index_previous"] - 1)
category_momentum = category_momentum[category_momentum["min_rows"] >= 500].sort_values("adjusted_momentum_pct", ascending=False)

print(f"Releases from {latest_year} vs {latest_year - 1}. Index > 1 means selling more than the average product at the same age and discount.")
fig, ax = plt.subplots(figsize=(10, 8))
plot_df = category_momentum.iloc[::-1]
y = range(len(plot_df))
ax.barh([i + 0.2 for i in y], plot_df["raw_growth_pct"], height=0.4, label="Raw sales growth", color="lightgray")
ax.barh([i - 0.2 for i in y], plot_df["adjusted_momentum_pct"], height=0.4, label="Discount- and age-adjusted momentum", color="steelblue")
ax.set_yticks(list(y), plot_df.index)
ax.axvline(0, color="black", linewidth=0.8)
ax.set_xlabel("% change, latest vs previous release year")
ax.set_title("Category momentum: raw vs discount- and age-adjusted")
ax.legend()
fig.tight_layout()
plt.show()

category_momentum.round(2)
```

**Summary: discount- and age-adjusted category momentum (2019 vs 2018 releases)**
- Adjusting for months since release and discount band changes the ranking. Rising: printed shirt (+48%), sheath dress (+11%), shorts (+8%), long dress (+7%). Falling: long duster (-32%), gitana skirt (-32%), maxi (-27%), trapeze dress (-25%).
- Doll dress, culottes and long sleeve grew 13-69% in raw sales but are flat after adjustment: more products were released, not more demand per product.
- Small categories (such as printed shirt) have noisy growth figures. Treat the ranking as exploratory.

## Weekly sales and restocks; purchase timing

Customer transactions have timestamps, so they can be aggregated into ISO weeks and aligned with the weekly restock records. The first chart shows the three highest-sales product-retailer pairs among those with at least 10 weeks containing both sales and restock records. The heatmap shows average daily purchase quantity by weekday and hour, including zero-purchase dates and hours.


```python
import matplotlib.pyplot as plt
import pandas as pd

# Align transaction sales and restocks using ISO calendar weeks.
purchase_datetime = pd.to_datetime(cus_df["data"], errors="raise")
purchase_iso = purchase_datetime.dt.isocalendar()
weekly_customer_df = cus_df[["external_code", "retail", "qty"]].copy()
weekly_customer_df["iso_year"] = purchase_iso.year.astype("int16")
weekly_customer_df["iso_week"] = purchase_iso.week.astype("int8")
weekly_keys = ["external_code", "retail", "iso_year", "iso_week"]

sales_by_week = (
    weekly_customer_df.groupby(weekly_keys, as_index=False)["qty"]
    .sum()
    .rename(columns={"qty": "sales_qty"})
)
restocks_by_week = (
    restocks_df.groupby(["external_code", "retail", "year", "week"], as_index=False)["qty"]
    .sum()
    .rename(columns={"year": "iso_year", "week": "iso_week", "qty": "restock_qty"})
)
aligned_weeks = sales_by_week.merge(restocks_by_week, on=weekly_keys, how="inner")
representative_pairs = (
    aligned_weeks.groupby(["external_code", "retail"], as_index=False)
    .agg(overlap_weeks=("iso_week", "size"), sales_qty=("sales_qty", "sum"))
)
representative_pairs = representative_pairs.loc[
    representative_pairs["overlap_weeks"] >= 10
].nlargest(3, "sales_qty")
if representative_pairs.empty:
    raise ValueError("No product-retailer pairs have at least 10 aligned sales/restock weeks.")

fig, axes = plt.subplots(
    len(representative_pairs), 1,
    figsize=(12, 4 * len(representative_pairs)),
    squeeze=False,
)
for ax, pair in zip(axes.flat, representative_pairs.itertuples(index=False)):
    pair_sales = sales_by_week.loc[
        (sales_by_week["external_code"] == pair.external_code)
        & (sales_by_week["retail"] == pair.retail)
    ].copy()
    pair_restocks = restocks_by_week.loc[
        (restocks_by_week["external_code"] == pair.external_code)
        & (restocks_by_week["retail"] == pair.retail)
    ].copy()
    for weekly_data in (pair_sales, pair_restocks):
        weekly_data["week_start"] = pd.to_datetime(
            weekly_data["iso_year"].astype(str)
            + "-W"
            + weekly_data["iso_week"].astype(str).str.zfill(2)
            + "-1",
            format="%G-W%V-%u",
        )
    pair_sales = pair_sales.set_index("week_start")["sales_qty"]
    pair_restocks = pair_restocks.set_index("week_start")["restock_qty"]
    weeks = pd.date_range(
        min(pair_sales.index.min(), pair_restocks.index.min()),
        max(pair_sales.index.max(), pair_restocks.index.max()),
        freq="W-MON",
    )
    sales_line = ax.plot(
        weeks, pair_sales.reindex(weeks, fill_value=0),
        color="tab:blue", marker=".", label="Sales",
    )
    ax.set_ylabel("Sales (units)", color="tab:blue")
    restock_ax = ax.twinx()
    restock_line = restock_ax.plot(
        weeks, pair_restocks.reindex(weeks, fill_value=0),
        color="tab:orange", marker=".", linestyle="--", label="Restocks",
    )
    restock_ax.set_ylabel("Restocks (units)", color="tab:orange")
    ax.set_title(
        f"Product {pair.external_code} at retailer {pair.retail} "
        f"({pair.overlap_weeks} overlapping weeks)"
    )
    ax.legend(sales_line + restock_line, ["Sales", "Restocks"], loc="upper right")
    ax.grid(axis="x", alpha=0.25)
axes[-1, 0].set_xlabel("Week (starting Monday)")
fig.tight_layout()
plt.show()

# Average daily quantity by weekday and hour; include dates/hours with no purchases.
purchase_time = pd.DataFrame({
    "date": purchase_datetime.dt.normalize(),
    "hour": purchase_datetime.dt.hour,
    "qty": cus_df["qty"],
})
daily_hourly_qty = (
    purchase_time.groupby(["date", "hour"])["qty"]
    .sum()
    .unstack("hour", fill_value=0)
)
all_dates = pd.date_range(
    purchase_datetime.min().normalize(),
    purchase_datetime.max().normalize(),
    freq="D",
)
daily_hourly_qty = daily_hourly_qty.reindex(
    index=all_dates, columns=range(24), fill_value=0
)
weekday_hourly_avg = daily_hourly_qty.groupby(
    daily_hourly_qty.index.dayofweek
).mean().reindex(range(7), fill_value=0)

fig, ax = plt.subplots(figsize=(12, 5))
image = ax.imshow(weekday_hourly_avg.to_numpy(), aspect="auto", cmap="YlOrRd")
ax.set_xticks(range(24), [f"{hour:02d}:00" for hour in range(24)], rotation=45)
ax.set_yticks(range(7), ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"])
ax.set_xlabel("Hour of day")
ax.set_ylabel("Day of week")
ax.set_title("Average daily purchase quantity by weekday and hour")
fig.colorbar(image, ax=ax, label="Average quantity per day")
fig.tight_layout()
plt.show()

representative_pairs
```

**Summary: weekly sales, restocks and purchase timing**
- Weekly sales and restocks for the highest-sales product-store pairs show how replenishment follows demand within a season.
- The weekday-by-hour heatmap shows when purchases cluster, which is useful for choosing the time grain of forecasts and for operational planning.
- Pairs were limited to those with enough overlapping weeks, so the examples are the best-covered pairs rather than a random sample.

## Image EDA

Each product has one image referenced by `sales_df["image_path"]`. This cell checks unique product images for missing or unreadable files, summarizes image dimensions and exact duplicates, and displays a reproducible contact sheet labeled with product metadata.


```python
from collections import Counter
from hashlib import sha256
from pathlib import Path

from PIL import Image

image_root = Path("data/visuelle2/images")
if not image_root.is_dir():
    image_root = Path("images")
if not image_root.is_dir():
    raise FileNotFoundError("Could not find the images directory (tried images/ and data/visuelle2/images/).")

product_images = sales_df[
    ["external_code", "category", "color", "season", "image_path"]
].drop_duplicates("external_code").copy()

image_records = []
missing_images = []
unreadable_images = []
image_hashes = {}
for row in product_images.itertuples(index=False):
    image_path = image_root / row.image_path
    if not image_path.is_file():
        missing_images.append(row.image_path)
        continue
    try:
        image_bytes = image_path.read_bytes()
        with Image.open(image_path) as image:
            image.verify()
        with Image.open(image_path) as image:
            width, height = image.size
            image_mode = image.mode
        digest = sha256(image_bytes).hexdigest()
        image_hashes.setdefault(digest, []).append(row.image_path)
        image_records.append({
            "external_code": row.external_code,
            "image_path": row.image_path,
            "width": width,
            "height": height,
            "mode": image_mode,
            "aspect_ratio": width / height,
        })
    except (OSError, ValueError, SyntaxError) as error:
        unreadable_images.append((row.image_path, str(error)))

image_stats = pd.DataFrame(image_records)
exact_duplicates = [paths for paths in image_hashes.values() if len(paths) > 1]
print(f"Products checked: {len(product_images):,}")
print(f"Missing images: {len(missing_images):,}")
print(f"Unreadable images: {len(unreadable_images):,}")
print(f"Exact duplicate image groups: {len(exact_duplicates):,}")
if not image_stats.empty:
    display(image_stats[["width", "height", "aspect_ratio"]].describe().round(2))
if missing_images:
    print("Missing image paths:", missing_images[:10])
if unreadable_images:
    print("Unreadable image examples:", unreadable_images[:10])

# Sample one row per product so repeated retailer records do not overrepresent products.
sample_products = product_images.sample(
    n=min(20, len(product_images)), random_state=42
)
columns = 5
rows = (len(sample_products) + columns - 1) // columns
fig, axes = plt.subplots(rows, columns, figsize=(15, 3.5 * rows), squeeze=False)
for ax in axes.flat:
    ax.axis("off")
for ax, row in zip(axes.flat, sample_products.itertuples(index=False)):
    image_path = image_root / row.image_path
    if not image_path.is_file():
        ax.set_title(f"Missing: {row.image_path}", fontsize=8)
        continue
    try:
        with Image.open(image_path) as image:
            rgba = image.convert("RGBA")
            white_background = Image.new("RGBA", rgba.size, "white")
            white_background.alpha_composite(rgba)
            ax.imshow(white_background.convert("RGB"))
        ax.set_title(
            f"{row.category} | {row.color} | {row.season}\n"
            f"Product {row.external_code}",
            fontsize=8,
        )
    except (OSError, ValueError, SyntaxError) as error:
        ax.set_title(f"Unreadable: {row.image_path}\n{error}", fontsize=7)
plt.tight_layout()
plt.show()
```

**Summary: image quality check**
- 5,355 products were checked: none had a missing image, 1 was unreadable (AI19/04442.png, a corrupt PNG) and 2 groups of images were exact duplicates.
- Image sizes vary (median about 577x728 px, aspect ratio 0.23-2.84), so images must be resized or padded before being used in a model.
- Image data is clean enough to use, after excluding the one unreadable file.

## One-month category-sales forecast

This section uses timestamped VISUELLE customer purchases as the target, maps purchased products to their catalog category, and forecasts the next calendar month's category sales. It is a retail-sales proxy for fashion trends, not a Google search-interest forecast. The forecast month is derived from the latest transaction month in the data (currently June 2021, so July 2021), not the current real-world month.

Compare last-month, trailing-three-month-average, and seasonal-naive baselines with a rolling-origin backtest over the latest 12 calendar months. The notebook selects the lowest-WAPE baseline, then shows category rankings and recent momentum. Check coverage and backtest results before interpreting the ranking: the data has calendar gaps and a sharp post-February-2020 decline, so this is exploratory rather than a reliable live forecast.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# The customer transaction timestamps provide calendar-month sales.
forecast_category_lookup = sales_df[["external_code", "category"]].drop_duplicates()
if forecast_category_lookup["external_code"].duplicated().any():
    raise ValueError("Each product must map to exactly one category before aggregation.")

forecast_transactions = cus_df[["external_code", "data", "qty"]].copy()
forecast_transactions["data"] = pd.to_datetime(forecast_transactions["data"], errors="raise")
forecast_transactions = forecast_transactions.merge(
    forecast_category_lookup,
    on="external_code",
    how="left",
    validate="many_to_one",
)
total_transaction_qty = forecast_transactions["qty"].sum()
mapped_transaction_qty = forecast_transactions.loc[
    forecast_transactions["category"].notna(), "qty"
].sum()
category_mapping_coverage = mapped_transaction_qty / total_transaction_qty

category_transactions = forecast_transactions.dropna(subset=["category"]).copy()
category_transactions["month"] = category_transactions["data"].dt.to_period("M")
monthly_category_sales = category_transactions.groupby(
    ["month", "category"], observed=True
)["qty"].sum().unstack(fill_value=0)
calendar_months = pd.period_range(
    monthly_category_sales.index.min(), monthly_category_sales.index.max(), freq="M"
)
monthly_category_sales = monthly_category_sales.reindex(calendar_months, fill_value=0)
monthly_category_sales.index.name = "month"
monthly_category_sales.columns.name = "category"

# Zero-fill months without transactions so the calendar remains continuous; report them as possible data gaps.
months_without_transactions = monthly_category_sales.index[
    monthly_category_sales.sum(axis=1).eq(0)
]
baseline_predictions = {
    "Last month": monthly_category_sales.shift(1),
    "Trailing 3-month average": monthly_category_sales.shift(1).rolling(
        window=3, min_periods=1
    ).mean(),
    "Seasonal naive (12 months ago)": monthly_category_sales.shift(12),
}
backtest_months = monthly_category_sales.index[-12:]
backtest_rows = []
for baseline_name, predictions in baseline_predictions.items():
    actual = monthly_category_sales.loc[backtest_months]
    predicted = predictions.loc[backtest_months]
    valid_months = predicted.notna().all(axis=1)
    actual_values = actual.loc[valid_months].to_numpy(dtype=float)
    predicted_values = predicted.loc[valid_months].to_numpy(dtype=float)
    actual_total = actual_values.sum()
    backtest_rows.append({
        "baseline": baseline_name,
        "WAPE %": 100 * np.abs(actual_values - predicted_values).sum() / actual_total,
        "MAE units per category-month": np.abs(actual_values - predicted_values).mean(),
        "months evaluated": int(valid_months.sum()),
    })

backtest_summary = pd.DataFrame(backtest_rows).sort_values("WAPE %").reset_index(drop=True)
best_baseline = backtest_summary.loc[0, "baseline"]
if best_baseline == "Last month":
    next_month_prediction = monthly_category_sales.iloc[-1].copy()
elif best_baseline == "Trailing 3-month average":
    next_month_prediction = monthly_category_sales.tail(3).mean()
else:
    next_month_prediction = monthly_category_sales.iloc[-12].copy()
next_month_prediction = next_month_prediction.clip(lower=0)
next_month = monthly_category_sales.index[-1] + 1

last_month_total = monthly_category_sales.iloc[-1].sum()
prior_year_totals = monthly_category_sales.sum(axis=1).iloc[-13:-1]
prior_year_active_totals = prior_year_totals[prior_year_totals > 0]
if len(prior_year_active_totals):
    prior_year_median = prior_year_active_totals.median()
    latest_to_prior_median = last_month_total / prior_year_median
else:
    prior_year_median = np.nan
    latest_to_prior_median = np.nan

recent_3m_average = monthly_category_sales.tail(3).mean()
previous_3m_average = monthly_category_sales.iloc[-6:-3].mean()
recent_momentum_pct = 100 * (
    (recent_3m_average - previous_3m_average) / previous_3m_average.replace(0, np.nan)
)
category_forecast = pd.DataFrame({
    "forecast_units": next_month_prediction,
    "last_month_units": monthly_category_sales.iloc[-1],
    "recent_3m_average": recent_3m_average,
    "recent_momentum_pct": recent_momentum_pct,
}).sort_values("forecast_units", ascending=False)

print(f"Calendar data: {monthly_category_sales.index[0]} to {monthly_category_sales.index[-1]}")
print(f"Forecast month: {next_month} | selected baseline: {best_baseline}")
print(f"Purchased units mapped to a catalog category: {category_mapping_coverage:.1%}")
print(f"Months with no recorded transactions: {list(months_without_transactions.astype(str))}")
if pd.notna(prior_year_median):
    print(
        f"Latest-month units: {last_month_total:,.0f}; "
        f"{latest_to_prior_median:.1%} of the prior 12-month median "
        f"({prior_year_median:,.0f} units)."
    )
    if latest_to_prior_median < 0.25:
        print("Warning: latest-month volume is below 25% of the prior-year median; treat this forecast as low-confidence due to a possible coverage or regime change.")
print(f"Best rolling-origin WAPE: {backtest_summary.iloc[0]['WAPE %']:.1f}%")
print("Rolling-origin backtest over the latest 12 calendar months:")
display(backtest_summary.round(2))
print(f"Top categories by forecast units for {next_month}:")
display(category_forecast.head(10).round(2))

monthly_total_sales = monthly_category_sales.sum(axis=1)
fig, ax = plt.subplots(figsize=(12, 5))
ax.plot(
    monthly_total_sales.index.to_timestamp(),
    monthly_total_sales.values,
    marker="o",
    label="Observed category-mapped sales",
)
ax.scatter(
    next_month.to_timestamp(),
    next_month_prediction.sum(),
    marker="X",
    s=100,
    color="tab:red",
    label=f"{next_month} forecast ({best_baseline})",
    zorder=3,
)
ax.set_title("VISUELLE monthly sales and one-month-ahead baseline forecast")
ax.set_xlabel("Month")
ax.set_ylabel("Purchased units")
ax.legend()
ax.grid(alpha=0.25)
fig.tight_layout()
plt.show()
```

**Summary: one-month category-sales forecast**
- The data runs from 2016-12 to 2021-06 and the target is July 2021. 98.6% of purchased units map to a catalog category.
- Last month's sales was the best baseline, but still had a WAPE of 94%. The seasonal-naive baseline was far worse (WAPE above 5,000%).
- Latest-month volume is only 5.5% of the prior-year median, and two months have no transactions, so this looks like a coverage or regime change. The forecast is low-confidence and exploratory, not a reliable live forecast.

## Does search interest improve sales forecasts?

Predict each new product-retailer pair's first 12 weekly unit sales on the official VISUELLE2 train/test split. Models are nested so each step adds one signal:

1. **Sales/stock baseline** – category, colour, fabric, season, retailer, release timing and the initial restock quantity.
2. **+ price and discounts** – launch price and the 12-week discount schedule.
3. **+ Google Trends (pre-release)** – category/colour/fabric search interest in the 12 weeks *before* release. This is the only search variant usable at forecast time.
4. **+ Google Trends (in-season)** – search interest during the sales window. It uses information unavailable at forecast time, so treat it as an upper bound.

Search signals are judged by the WAPE change relative to the price/discount model, with a paired bootstrap over products so the difference is not just noise.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from xgboost import XGBRegressor

DATA = "data/visuelle2/"
weeks = [str(i) for i in range(12)]
scale = float(np.load(DATA + "stfore_sales_norm_scalar.npy"))

train = pd.read_csv(DATA + "stfore_train.csv")
test = pd.read_csv(DATA + "stfore_test.csv")
disc = pd.read_csv(DATA + "price_discount_series.csv").rename(columns={w: f"disc_{w}" for w in weeks})
trends = pd.read_csv(DATA + "vis2_gtrends_data.csv", parse_dates=["date"]).set_index("date")
trends = trends.where(trends >= 0)  # -1 marks missing

df = pd.concat([train.assign(split="train"), test.assign(split="test")], ignore_index=True)
df["release_date"] = pd.to_datetime(df["release_date"])
df = df.merge(disc, on=["external_code", "retail"], how="left")
is_train = (df["split"] == "train").to_numpy()
y = df[weeks].to_numpy() * scale  # back to unit sales

# Sales/stock baseline: catalog metadata, retailer, timing and initial restock.
base = pd.get_dummies(df[["category", "color", "fabric", "season"]], dtype=float)
base["retail"] = df["retail"]
base["restock"] = df["restock"]
base["week_of_year"] = df["release_date"].dt.isocalendar().week.astype(int)
base["month"] = df["release_date"].dt.month

price = df[["price"] + [f"disc_{w}" for w in weeks]].copy()
price["disc_mean"] = price[[f"disc_{w}" for w in weeks]].mean(axis=1)

release_pos = trends.index.get_indexer(df["release_date"], method="nearest")


def search_block(offsets, prefix):
    """Search interest for each product's category, colour and fabric at weekly offsets from release."""
    out = {}
    for attr in ["category", "color", "fabric"]:
        values = df[attr].to_numpy()
        for off in offsets:
            pos = np.clip(release_pos + off, 0, len(trends) - 1)
            col = np.full(len(df), np.nan)
            for term in np.unique(values):
                if term in trends.columns:
                    mask = values == term
                    col[mask] = trends[term].to_numpy()[pos[mask]]
            out[f"{prefix}_{attr}_{off:+d}w"] = col
    return pd.DataFrame(out)


search_pre = search_block(range(-12, 0, 2), "pre")
search_in = search_block(range(0, 12, 2), "in")

feature_sets = {
    "Sales/stock baseline": [base],
    "+ price & discounts": [base, price],
    "+ Google Trends (pre-release)": [base, price, search_pre],
    "+ Google Trends (in-season, leaky)": [base, price, search_in],
}

y_test = y[~is_train]
products_test = df.loc[~is_train, "external_code"].to_numpy()
predictions = {}
for name, parts in feature_sets.items():
    X = pd.concat(parts, axis=1)
    X.columns = [str(c) for c in X.columns]
    model = XGBRegressor(
        n_estimators=120, max_depth=5, learning_rate=0.1, subsample=0.8,
        colsample_bytree=0.8, tree_method="hist", max_bin=64,
        multi_strategy="one_output_per_tree", n_jobs=-1, random_state=0,
    )
    model.fit(X[is_train], y[is_train])
    predictions[name] = np.clip(model.predict(X[~is_train]), 0, None)

# Paired bootstrap over products: resample whole products so retailers of one item stay together.
codes, inverse = np.unique(products_test, return_inverse=True)
abs_err = {name: np.bincount(inverse, weights=np.abs(y_test - p).sum(axis=1)) for name, p in predictions.items()}
actual = np.bincount(inverse, weights=y_test.sum(axis=1))
rng = np.random.default_rng(0)
draws = rng.integers(0, len(codes), size=(1000, len(codes)))


def wape(err, idx=slice(None)):
    return 100 * err[idx].sum() / actual[idx].sum()


reference = "+ price & discounts"
rows = []
for name in feature_sets:
    boot = np.array([wape(abs_err[name], d) - wape(abs_err[reference], d) for d in draws])
    rows.append({
        "model": name,
        "WAPE %": wape(abs_err[name]),
        "MAE units/week": np.abs(y_test - predictions[name]).mean(),
        "WAPE change vs price model (pp)": wape(abs_err[name]) - wape(abs_err[reference]),
        "95% CI low": np.percentile(boot, 2.5),
        "95% CI high": np.percentile(boot, 97.5),
    })

search_results = pd.DataFrame(rows).set_index("model")
display(search_results.round(2))

ax = search_results["WAPE %"].plot(kind="barh", figsize=(9, 4), color="steelblue")
ax.set_xlabel("Test WAPE % (lower is better)")
ax.set_ylabel("")
ax.set_title("Does adding Google Trends improve 12-week sales forecasts?")
ax.invert_yaxis()
plt.tight_layout()
plt.show()

pre_row = search_results.loc["+ Google Trends (pre-release)"]
if pre_row["95% CI high"] < 0:
    print("Pre-release search interest gives a statistically reliable improvement over price/discount.")
elif pre_row["95% CI low"] > 0:
    print("Pre-release search interest makes forecasts reliably worse than price/discount alone.")
else:
    print("Pre-release search interest shows no reliable improvement: the 95% CI includes zero.")
```

**Summary: does search interest improve new-product forecasts?**
- On the official train/test split, adding price and discounts cut WAPE from 76.1% to 66.0%.
- Adding pre-release Google Trends made the forecast slightly worse (66.3%; +0.32 pp, 95% CI 0.04 to 0.64). Even the in-season variant, which leaks future information, did not help (66.4%).
- Price and discount carry the useful signal. Search interest does not improve these sales forecasts.

```python
import numpy as np
import pandas as pd
from sklearn.ensemble import HistGradientBoostingRegressor

# Test whether the search-interest signals shipped with VISUELLE improve monthly sales forecasts.
keys = ["external_code", "retail"]
months = [str(m) for m in range(12)]

units = sales_df.melt(
    id_vars=keys + ["category", "color", "fabric", "release_date", "restock"],
    value_vars=months, var_name="month", value_name="units",
)
disc = pds_df.melt(id_vars=keys + ["price"], value_vars=months, var_name="month", value_name="discount")
panel = units.merge(disc, on=keys + ["month"], how="inner", validate="one_to_one")
panel["month"] = panel["month"].astype(int)
panel["release_date"] = pd.to_datetime(panel["release_date"])
panel["month_start"] = panel["release_date"] + pd.to_timedelta((panel["month"] * 30.44).round(), unit="D")
panel = panel.rename(columns={"restock": "initial_stock"})

# Restocks received during each month of a product-retailer's life.
rs = restocks_df[keys + ["year", "week", "qty"]].merge(sales_df[keys + ["release_date"]], on=keys)
# ISO week/year of each restock, so the cell does not depend on an earlier cell adding a date column.
rs["restock_date"] = pd.to_datetime(
    rs["year"].astype(str) + "-" + rs["week"].astype(str) + "-1", format="%G-%V-%u", errors="coerce"
)
rs = rs.dropna(subset=["restock_date"])
rs["month"] = ((rs["restock_date"] - pd.to_datetime(rs["release_date"])).dt.days // 30.44).astype(int)
rs = rs[rs["month"].between(0, 11)].groupby(keys + ["month"], as_index=False)["qty"].sum()
panel = panel.merge(rs.rename(columns={"qty": "restock_qty"}), on=keys + ["month"], how="left")
panel["restock_qty"] = panel["restock_qty"].fillna(0)

# Search interest known at the start of each month (latest weekly value on or before that date).
trends = vis_gtrends_df.copy()
trends["date"] = pd.to_datetime(trends["date"])
trends = trends.sort_values("date")
panel = panel.sort_values("month_start")
for attr in ["category", "color", "fabric"]:
    long = trends[["date"] + list(trends.columns[1:])].melt(id_vars="date", var_name=attr, value_name=f"search_{attr}")
    long[f"search_{attr}"] = long[f"search_{attr}"].where(long[f"search_{attr}"] >= 0)  # -1 marks missing
    panel = pd.merge_asof(
        panel, long.sort_values("date"), left_on="month_start", right_on="date",
        by=attr, direction="backward", tolerance=pd.Timedelta(days=7),
    ).drop(columns="date")
search_cols = ["search_category", "search_color", "search_fabric"]

panel = panel.sort_values(keys + ["month"]).reset_index(drop=True)
panel["units_lag1"] = panel.groupby(keys)["units"].shift(1)
panel["units_lag2"] = panel.groupby(keys)["units"].shift(2)
model_df = panel.dropna(subset=["units_lag1", "units_lag2"] + search_cols).copy()
for c in ["category", "color", "fabric"]:
    model_df[c] = model_df[c].astype("category")

base_features = ["units_lag1", "units_lag2", "month", "price", "discount", "initial_stock", "restock_qty", "retail"]
search_features = base_features + search_cols

# Time-based split: train on earlier releases, test on the latest 25% of releases.
cutoff = model_df["release_date"].quantile(0.75)
train = model_df[model_df["release_date"] <= cutoff]
test = model_df[model_df["release_date"] > cutoff].copy()

def fit_predict(features):
    model = HistGradientBoostingRegressor(max_iter=300, learning_rate=0.05, random_state=0)
    model.fit(train[features], train["units"])
    return model.predict(test[features])

test["pred_last_month"] = test["units_lag1"]
test["pred_no_search"] = fit_predict(base_features)
test["pred_with_search"] = fit_predict(search_features)

def wape(df, col):
    return 100 * (df["units"] - df[col]).abs().sum() / df["units"].sum()

results = pd.DataFrame({
    "model": ["Last month's sales", "Sales + stock + price + discount", "Same + search interest"],
    "WAPE %": [wape(test, c) for c in ["pred_last_month", "pred_no_search", "pred_with_search"]],
    "MAE": [(test["units"] - test[c]).abs().mean() for c in ["pred_last_month", "pred_no_search", "pred_with_search"]],
})
print(f"Train rows: {len(train):,} | test rows: {len(test):,} (releases after {cutoff.date()})")
display(results.round(3))

# Paired bootstrap over products: does adding search reduce absolute error?
err_gain = (test["units"] - test["pred_no_search"]).abs() - (test["units"] - test["pred_with_search"]).abs()
by_product = err_gain.groupby(test["external_code"]).mean()
rng = np.random.default_rng(0)
boot = [by_product.sample(len(by_product), replace=True, random_state=int(s)).mean() for s in rng.integers(0, 1e9, 1000)]
low, high = np.percentile(boot, [2.5, 97.5])
print(f"Mean absolute-error reduction from search: {by_product.mean():.4f} units (95% CI {low:.4f} to {high:.4f})")
print("Search helps" if low > 0 else "No reliable improvement from search interest" if high >= 0 else "Search hurts")

# Do search signals relate to sales beyond what last month's sales already explain?
resid = model_df["units"] - model_df["units_lag1"]
print(model_df[search_cols].corrwith(resid).round(3).rename("corr with month-over-month change"))
```

## Data preparation and preprocessing pipeline

Steps follow the written pipeline: (1) integrate tables, (2) clean, (3) handle missing values, (4) handle outliers, (5) engineer features, (6) select features, (7) validation design, (8) synthetic data. Every join is anchored to the sales catalog, and the analysis window is 2017-2019 because customer data collapses from March 2020. Objects prefixed `prep_` are the pipeline outputs. Run after the loading cells above (`sales_df`, `pds_df`, `cus_df`, `restocks_df`, `vis_gtrends_df`).

### 1. Data integration

Sales and price/discount share exactly the same (product, retailer) keys, so the 1:1 join loses no rows. Restocks are aggregated to months since release, and search interest is matched with an as-of join (latest weekly value on or before the date) so nothing from the future leaks in. The `-1` placeholders in the search table are converted to missing. Weather is not used.

```python
import numpy as np
import pandas as pd

keys = ["external_code", "retail"]
month_cols = [str(m) for m in range(12)]

# 1a. Sales + price/discount: one row per product-retailer-month.
assert not sales_df.duplicated(keys).any(), "catalog keys must be unique"
units_long = sales_df.melt(
    id_vars=keys + ["category", "color", "fabric", "season", "release_date", "restock"],
    value_vars=month_cols, var_name="month", value_name="units",
)
discount_long = pds_df.melt(
    id_vars=keys + ["price"], value_vars=month_cols, var_name="month", value_name="discount",
)
prep_panel = units_long.merge(discount_long, on=keys + ["month"], how="inner", validate="one_to_one")
prep_panel["month"] = prep_panel["month"].astype(int)
prep_panel["release_date"] = pd.to_datetime(prep_panel["release_date"])
prep_panel["month_start"] = prep_panel["release_date"] + pd.to_timedelta(
    (prep_panel["month"] * 30.44).round(), unit="D"
)
prep_panel = prep_panel.rename(columns={"restock": "initial_stock"})
print(f"1a. Panel rows: {len(prep_panel):,} (sales rows x 12 months = {len(sales_df) * 12:,})")

# 1b. Restocks: ISO week to date, then summed by month since release.
restock_pairs = restocks_df[keys].drop_duplicates()
orphan_restock_pairs = len(
    restock_pairs.merge(sales_df[keys], on=keys, how="left", indicator=True).query("_merge == 'left_only'")
)
restocks = restocks_df.merge(sales_df[keys + ["release_date"]], on=keys, how="inner", validate="many_to_one")
restocks["restock_date"] = pd.to_datetime(
    restocks["year"].astype(str) + "-" + restocks["week"].astype(str) + "-1",
    format="%G-%V-%u", errors="coerce",
)
restocks = restocks.dropna(subset=["restock_date"])
restocks["month"] = (
    (restocks["restock_date"] - pd.to_datetime(restocks["release_date"])).dt.days // 30.44
).astype(int)
restocks_monthly = (
    restocks[restocks["month"].between(0, 11)]
    .groupby(keys + ["month"], as_index=False)["qty"].sum()
    .rename(columns={"qty": "restock_qty"})
)
prep_panel = prep_panel.merge(restocks_monthly, on=keys + ["month"], how="left", validate="many_to_one")
print(f"1b. Restock pairs not in the sales catalog (dropped by the inner join): {orphan_restock_pairs:,}")

# 1c. Search interest from the weekly keyword table, joined on category, color and fabric.
trends = vis_gtrends_df.copy()
trends["date"] = pd.to_datetime(trends["date"])
trends = trends.sort_values("date")
search_cols = []
pre_release_search = {}
prep_panel = prep_panel.sort_values("month_start")
for attr in ["category", "color", "fabric"]:
    keywords = [c for c in trends.columns if c in set(sales_df[attr])]
    long = trends[["date"] + keywords].melt(id_vars="date", var_name=attr, value_name="value")
    long["value"] = long["value"].where(long["value"] >= 0)  # -1 marks a missing value
    long = long.sort_values("date")
    # Latest weekly value known at the start of each month.
    current = long.rename(columns={"value": f"search_{attr}"})
    prep_panel = pd.merge_asof(
        prep_panel, current, left_on="month_start", right_on="date",
        by=attr, direction="backward", tolerance=pd.Timedelta(days=7),
    ).drop(columns="date")
    # Mean of the 12 weeks before release, usable for brand-new products.
    long[f"search_pre_{attr}"] = long.groupby(attr)["value"].transform(
        lambda s: s.rolling(12, min_periods=6).mean().shift(1)
    )
    search_cols += [f"search_{attr}", f"search_pre_{attr}"]
    pre_release_search[attr] = long[["date", attr, f"search_pre_{attr}"]]
prep_panel = prep_panel.sort_values("release_date")
for attr in ["category", "color", "fabric"]:
    prep_panel = pd.merge_asof(
        prep_panel, pre_release_search[attr].sort_values("date"), left_on="release_date", right_on="date",
        by=attr, direction="backward", tolerance=pd.Timedelta(days=7),
    ).drop(columns="date")
prep_panel = prep_panel.sort_values(keys + ["month"]).reset_index(drop=True)
print(f"1c. Integrated panel: {prep_panel.shape[0]:,} rows x {prep_panel.shape[1]} columns")
display(prep_panel.head())
```

### 2. Data cleaning

Orphan customer rows (products not in the catalog) are dropped by joining on the catalog, the single duplicate customer row is removed, dates are parsed, and the analysis window is restricted to 2017-2019. Customer transactions are aggregated to calendar months per category, which is the target for the category trend analysis.

```python
ANALYSIS_START, ANALYSIS_END = pd.Timestamp("2017-01-01"), pd.Timestamp("2019-12-31")

# Customer transactions: remove duplicates and orphans, then aggregate to category-months.
catalog_categories = sales_df[["external_code", "category"]].drop_duplicates("external_code")
customers = cus_df.drop_duplicates()
duplicates_removed = len(cus_df) - len(customers)
customers = customers.merge(catalog_categories, on="external_code", how="inner", validate="many_to_one")
orphans_removed = len(cus_df) - duplicates_removed - len(customers)
customers["date"] = pd.to_datetime(customers["data"], errors="raise")
customers = customers[customers["date"].between(ANALYSIS_START, ANALYSIS_END + pd.Timedelta(days=1))]
prep_category_monthly = (
    customers.assign(month=customers["date"].dt.to_period("M"))
    .groupby(["month", "category"])["qty"].sum().unstack(fill_value=0)
)

# Panel: drop duplicates and keep the analysis window.
rows_before = len(prep_panel)
prep_panel = prep_panel.drop_duplicates(keys + ["month"])
prep_panel = prep_panel[prep_panel["month_start"].between(ANALYSIS_START, ANALYSIS_END)].reset_index(drop=True)

print(f"Customer duplicate rows removed: {duplicates_removed:,}")
print(f"Customer rows with products not in the catalog removed: {orphans_removed:,}")
print(f"Customer months kept: {prep_category_monthly.index.min()} to {prep_category_monthly.index.max()}")
print(f"Panel rows kept in {ANALYSIS_START.year}-{ANALYSIS_END.year}: {len(prep_panel):,} of {rows_before:,}")
```

### 3. Missing values

The single missing discount is dropped, a missing restock means no restock so it becomes 0, and missing search values are left as missing because the `-1` placeholder is not a real value and imputing it would invent signal. Gradient-boosted trees handle the remaining gaps natively.

```python
missing_before = prep_panel.isna().sum()

prep_panel = prep_panel.dropna(subset=["units", "price", "discount"])
prep_panel["restock_qty"] = prep_panel["restock_qty"].fillna(0)

missing_after = prep_panel.isna().sum()
missing_report = pd.DataFrame({
    "missing_before": missing_before,
    "missing_after": missing_after,
    "missing_after_%": (100 * missing_after / len(prep_panel)).round(2),
})
print(f"Rows after dropping missing units/price/discount: {len(prep_panel):,}")
display(missing_report[(missing_report["missing_before"] > 0) | (missing_report["missing_after"] > 0)])
```

### 4. Outliers

Negative monthly sales are treated as returns and kept, with a clipped column to test sensitivity. Heavy-tailed sales are not removed because the high sellers are the hits a trend model must learn. Categories with too few rows are flagged so that rankings are not driven by noise.

```python
MIN_CATEGORY_ROWS = 20000  # panel rows needed for a category to be ranked or modelled alone

negative_units = int((prep_panel["units"] < 0).sum())
prep_panel["units_clipped"] = prep_panel["units"].clip(lower=0)  # sensitivity check for returns

positive_units = prep_panel["units"].clip(lower=0)
top_1pct_share = positive_units.nlargest(int(len(positive_units) * 0.01)).sum() / positive_units.sum()
print(f"Negative monthly sales rows (kept as returns): {negative_units:,}")
print(f"Units quantiles: {prep_panel['units'].quantile([0.5, 0.9, 0.99, 0.999]).to_dict()}, max {prep_panel['units'].max():.0f}")
print(f"Share of units from the top 1% of rows: {top_1pct_share:.1%}")

category_rows = prep_panel.groupby("category").size().sort_values()
prep_small_categories = category_rows[category_rows < MIN_CATEGORY_ROWS].index.tolist()
print(f"Categories under {MIN_CATEGORY_ROWS:,} panel rows (flagged, not removed): {prep_small_categories}")
```

### 5. Feature engineering

Features cover sales history (lags and a trailing average, built only from past months), lifecycle (months since release, season, calendar month), price and discount, stock, search interest (value at the start of each month and the 12-week pre-release mean) and catalog attributes. Rows without enough history for the lags are dropped.

```python
prep_panel = prep_panel.sort_values(keys + ["month"]).reset_index(drop=True)

# Sales history from previous months only.
grouped_units = prep_panel.groupby(keys)["units"]
for lag in (1, 2, 3):
    prep_panel[f"units_lag{lag}"] = grouped_units.shift(lag)
prep_panel["units_avg_prev3"] = prep_panel[["units_lag1", "units_lag2", "units_lag3"]].mean(axis=1)

# Lifecycle and calendar.
prep_panel["calendar_month"] = prep_panel["month_start"].dt.month
prep_panel["release_month"] = prep_panel["release_date"].dt.month

# Price and discount.
prep_panel["discount_band"] = pd.cut(
    prep_panel["discount"], bins=[-0.001, 0, 0.1, 0.2, 0.3, 0.5, 1.0],
    labels=["0%", "0-10%", "10-20%", "20-30%", "30-50%", "50-100%"],
)

# Rows need two months of history for the lags.
prep_model_df = prep_panel.dropna(subset=["units_lag1", "units_lag2"]).copy()
for column in ["category", "color", "fabric", "season", "discount_band"]:
    prep_model_df[column] = prep_model_df[column].astype("category")

feature_groups = {
    "sales_history": ["units_lag1", "units_lag2", "units_avg_prev3"],
    "lifecycle": ["month", "calendar_month", "release_month", "season"],
    "price_discount": ["price", "discount"],
    "stock": ["initial_stock", "restock_qty"],
    "attributes": ["retail", "category", "color", "fabric"],
    "search_interest": ["search_category", "search_color", "search_fabric"],
    "search_pre_release": ["search_pre_category", "search_pre_color", "search_pre_fabric"],
}
print(f"Modelling rows: {len(prep_model_df):,} (dropped {len(prep_panel) - len(prep_model_df):,} without lag history)")
print({group: len(columns) for group, columns in feature_groups.items()})
display(prep_model_df[sum(feature_groups.values(), []) + ["units"]].describe().T.round(3))
```

### 6. Feature selection

Selection is domain-driven first, then checked with data. The time split is defined here so importance is measured on held-out rows. Search features are tested as an added group on top of sales, stock, price and discount. In-season search values (the weeks after a product's month) use information unavailable at forecast time, so they are not built into the main model.

```python
from sklearn.ensemble import HistGradientBoostingRegressor
from sklearn.inspection import permutation_importance

# Time-based split: train on earlier releases, test on the latest 25% of releases.
split_date = prep_model_df["release_date"].quantile(0.75)
prep_train = prep_model_df[prep_model_df["release_date"] <= split_date]
prep_test = prep_model_df[prep_model_df["release_date"] > split_date]

core_features = (
    feature_groups["sales_history"] + feature_groups["lifecycle"]
    + feature_groups["price_discount"] + feature_groups["stock"] + ["retail", "category"]
)
candidate_features = core_features + feature_groups["search_interest"]

# Correlation of search with month-over-month change beyond last month's sales.
change = prep_model_df["units"] - prep_model_df["units_lag1"]
print("Correlation of search features with month-over-month change:")
display(prep_model_df[feature_groups["search_interest"] + feature_groups["search_pre_release"]].corrwith(change).round(3).to_frame("corr"))

# Permutation importance on held-out rows.
selection_model = HistGradientBoostingRegressor(max_iter=150, learning_rate=0.08, random_state=0)
selection_model.fit(prep_train[candidate_features], prep_train["units"])
test_sample = prep_test.sample(20000, random_state=0)
importance = permutation_importance(
    selection_model, test_sample[candidate_features], test_sample["units"],
    scoring="neg_mean_absolute_error", n_repeats=3, random_state=0,
)
prep_feature_importance = (
    pd.Series(importance.importances_mean, index=candidate_features, name="MAE increase when shuffled")
    .sort_values(ascending=False)
)
display(prep_feature_importance.round(4).to_frame())
```

### 7. Validation design

The split is chronological with no shuffling, and a paired bootstrap over products tests whether adding a feature group changes the error by more than noise. The helpers below are reused for any model comparison.

```python
def wape(actual, predicted):
    return 100 * np.abs(actual - predicted).sum() / np.abs(actual).sum()


def paired_bootstrap(actual, baseline_pred, new_pred, groups, n_boot=1000, seed=0):
    """Mean absolute-error reduction of new_pred vs baseline_pred, resampling products."""
    gain = np.abs(actual - baseline_pred) - np.abs(actual - new_pred)
    per_group = gain.groupby(groups).mean()
    rng = np.random.default_rng(seed)
    boot = [per_group.sample(len(per_group), replace=True, random_state=int(s)).mean()
            for s in rng.integers(0, 1_000_000_000, n_boot)]
    low, high = np.percentile(boot, [2.5, 97.5])
    return per_group.mean(), low, high


print(f"Split date: releases after {split_date.date()} are the test set")
print(f"Train: {len(prep_train):,} rows, {prep_train['external_code'].nunique():,} products")
print(f"Test: {len(prep_test):,} rows, {prep_test['external_code'].nunique():,} products")
print(f"Baseline (last month's sales) WAPE on test: {wape(prep_test['units'], prep_test['units_lag1']):.1f}%")
```

### 8. Synthetic data

Synthetic data is not used in the main pipeline: there are over 100,000 product-store pairs, invented sales could bake in trend assumptions, and search interest cannot be fabricated without circular reasoning. The helper below bootstraps small categories in the training set only, as an optional sensitivity check. It is switched off by default and never touches the test set.

```python
USE_OVERSAMPLING = False
OVERSAMPLE_TARGET_ROWS = 10000


def oversample_small_categories(train_df, target_rows, seed=0):
    """Bootstrap rows of categories below target_rows (training data only)."""
    parts = [train_df]
    for category, count in train_df["category"].value_counts().items():
        if 0 < count < target_rows:
            parts.append(train_df[train_df["category"] == category].sample(
                target_rows - count, replace=True, random_state=seed))
    return pd.concat(parts, ignore_index=True)


train_counts = prep_train["category"].value_counts()
affected = train_counts[train_counts < OVERSAMPLE_TARGET_ROWS]
print(f"Categories that would be oversampled (target {OVERSAMPLE_TARGET_ROWS:,} rows): {affected.to_dict()}")
prep_train_final = oversample_small_categories(prep_train, OVERSAMPLE_TARGET_ROWS) if USE_OVERSAMPLING else prep_train
print(f"Oversampling enabled: {USE_OVERSAMPLING} | final training rows: {len(prep_train_final):,}")
```

## Do searches come before sales? (calendar-time weekly series)

The cells above forecast the 12-week sales of each *new* product (the optional extension in the proposal). This section is the main experiment:

1. Each product has only 12 weeks of sales, so listing-weeks are placed on the **calendar** (`release_date + 7 * week`) and **aggregated by category, colour and fabric** into weekly series.
2. **Out-of-stock weeks are excluded** because zero sales there do not reflect demand. In `sales.csv`, `restock` is the *total* quantity delivered over the whole record (it equals the sum of all rows in `restocks.csv` for 99.99% of product-retailer pairs and is never smaller than the sum of the rows dated up to release). It is therefore **not** extra launch stock on top of `restocks.csv`; adding the two double-counts. Stock is rebuilt from the dated restock rows only: rows dated before or in the release week form the opening stock (the 86,784 pre-release rows are goods that arrived before launch, not week-0 deliveries), later rows add stock in their own week, and sales deplete it. A week with no stock available at its start is dropped. The target is the mean units per in-stock listing, and errors are weighted by the number of in-stock listings (i.e. WAPE on unit totals).
3. **Split by season:** train = SS17, AW17, SS18, AW18; validation = SS19; test = AW19. Seasons overlap in calendar time, so each split's series is built only from that split's products.
4. **Search inputs:** the Google Trends series bundled with VISUELLE for each tag, shifted by 0 to 8 weeks (search at week *t - lag* predicts sales at week *t*), plus our own keyword series (`raw_pulls/trends_weekly_clean.csv`) for the tags that have a matching keyword.
5. The lag is chosen on validation and reported on test, with a paired bootstrap over test weeks against the model without search.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from xgboost import XGBRegressor

DATA = "data/visuelle2/"
```

```python
E_SPLITS = {"train": ["SS17", "AW17", "SS18", "AW18"], "val": ["SS19"], "test": ["AW19"]}
E_ATTRS = ["category", "color", "fabric"]
E_WEEKS = [str(i) for i in range(12)]

e3 = pd.read_csv(DATA + "sales.csv", index_col=0, parse_dates=["release_date"]).reset_index(drop=True)
e3["split"] = e3["season"].map({s: k for k, seasons in E_SPLITS.items() for s in seasons})
e3["lid"] = np.arange(len(e3))
display(e3.groupby(["split", "season"])["release_date"].agg(["min", "max", "count"]))

# `restock` equals the sum of ALL rows of restocks.csv, so it is not added to them. Stock is rebuilt from dated rows only.
S = e3[E_WEEKS].to_numpy(float)
rs = pd.read_csv(DATA + "restocks.csv")
rs["date"] = pd.to_datetime(rs["year"].astype(str) + "-W" + rs["week"].astype(str).str.zfill(2) + "-1", format="%G-W%V-%u")
rs = rs.merge(e3[["external_code", "retail", "release_date", "lid"]], on=["external_code", "retail"])
rs["k"] = (rs["date"] - rs["release_date"]).dt.days // 7  # weeks since release; negative = delivered before launch

all_rows = rs.groupby("lid")["qty"].sum().reindex(e3["lid"]).fillna(0).to_numpy()
up_to_release = rs[rs["k"] <= 0].groupby("lid")["qty"].sum().reindex(e3["lid"]).fillna(0).to_numpy()
restock_col = e3["restock"].to_numpy(float)
print(f"restock == sum of all restocks.csv rows: {np.mean(restock_col == all_rows):.4f}; "
      f"restock < sum of rows dated up to release: {np.mean(restock_col < up_to_release):.4f}")

R = np.zeros_like(S)
inwin = rs[(rs["k"] >= 1) & (rs["k"] < 12)]
np.add.at(R, (inwin["lid"].to_numpy(), inwin["k"].to_numpy(int)), inwin["qty"].to_numpy(float))
opening = up_to_release.copy()
no_rows = all_rows == 0
opening[no_rows] = restock_col[no_rows]  # pairs without any restock row: fall back to the column
print(f"pre-release rows: {(rs['k'] < 0).sum()} ({rs.loc[rs['k'] < 0, 'lid'].nunique()} listings) folded into opening stock; "
      f"{no_rows.sum()} listings fall back to `restock`")

# Stock available during week k = opening + restocks dated 1..k - sales of earlier weeks.
prior_sales = np.cumsum(S, axis=1) - S
stock_start = opening[:, None] + np.cumsum(R, axis=1) - prior_sales
in_stock = stock_start > 0
double_count = restock_col[:, None] + opening[:, None] + np.cumsum(R, axis=1) - prior_sales > 0  # wrong; shown for comparison
print(f"Listing-weeks excluded as out of stock: {(~in_stock).sum()} ({100 * (~in_stock).mean():.2f}%); "
      f"of these {(S[~in_stock] > 0).sum()} still show sales (stock data is only a proxy). "
      f"Adding `restock` on top of the rows would exclude just {100 * (~double_count).mean():.2f}%.")

pds = pd.read_csv(DATA + "price_discount_series.csv")
meta = e3[["external_code", "retail", "lid"]].merge(pds, on=["external_code", "retail"], how="left").sort_values("lid")
D = meta[E_WEEKS].to_numpy(float)

n_list = len(e3)
long = pd.DataFrame({
    "sales": S.ravel(),
    "in_stock": in_stock.ravel(),
    "disc": D.ravel(),
    "age": np.tile(np.arange(12), n_list),
    "price": np.repeat(meta["price"].to_numpy(float), 12),
    "split": np.repeat(e3["split"].to_numpy(), 12),
})
long["date"] = pd.DatetimeIndex(np.repeat(e3["release_date"].to_numpy(), 12)) + pd.to_timedelta(7 * long["age"], unit="D")
for attr in E_ATTRS:
    long[attr] = np.repeat(e3[attr].to_numpy(), 12)

# Weekly series per category / colour / fabric value, built separately for each split.
frames = []
for attr in E_ATTRS:
    keys = [attr, "split", "date"]
    active = long.groupby(keys).agg(n_active=("sales", "size"), age=("age", "mean"), disc=("disc", "mean"), price=("price", "mean"))
    stocked = long[long["in_stock"]].groupby(keys).agg(rate=("sales", "mean"), n_in=("sales", "size"))
    g = stocked.join(active).reset_index().rename(columns={attr: "term"})
    g["attr"] = attr
    frames.append(g)
panel = pd.concat(frames, ignore_index=True)
panel["week"] = panel["date"].dt.isocalendar().week.astype(int)
panel["month"] = panel["date"].dt.month
panel["attr_code"] = panel["attr"].astype("category").cat.codes
panel["term_code"] = (panel["attr"] + ":" + panel["term"]).astype("category").cat.codes
print(panel.groupby("split").size().to_dict(), "weekly series rows")
panel.head()
```

```python
LAGS = range(0, 9)

vis = pd.read_csv(DATA + "vis2_gtrends_data.csv", parse_dates=["date"]).set_index("date")
vis = vis.where(vis >= 0)  # -1 marks missing
vis = vis.reindex(pd.date_range(vis.index.min(), vis.index.max(), freq="7D"))

# Our own keyword series that match a VISUELLE tag (VISUELLE tag -> our keyword). Our weeks start on Sunday.
OWN_MAP = {
    "capris": "capri pants", "culottes": "culottes", "miniskirt": "mini skirt", "doll dress": "babydoll dress",
    "lace": "lace", "linen": "linen", "velvet": "velvet",
}
own_raw = pd.read_csv("raw_pulls/trends_weekly_clean.csv", index_col=0, parse_dates=True)
own_raw.index = own_raw.index + pd.Timedelta(days=1)
own = pd.DataFrame({tag: own_raw[kw] for tag, kw in OWN_MAP.items()}).reindex(vis.index)

tpos = vis.index.get_indexer(panel["date"])


def lagged_search(frame, lag):
    """Search interest of each row's term `lag` weeks before the sales week (NaN if the term has no series)."""
    shifted = frame.shift(lag)
    out = np.full(len(panel), np.nan)
    for term in frame.columns:
        m = (panel["term"] == term).to_numpy() & (tpos >= 0)
        if m.any():
            out[m] = shifted[term].to_numpy()[tpos[m]]
    return out


vis_lag = {lag: lagged_search(vis, lag) for lag in LAGS}
own_lag = {lag: lagged_search(own, lag) for lag in LAGS}

base_cols = ["attr_code", "term_code", "n_active", "age", "disc", "price", "week", "month"]
rate = panel["rate"].to_numpy()
weight = panel["n_in"].to_numpy(float)
tr = (panel["split"] == "train").to_numpy()
va = (panel["split"] == "val").to_numpy()
te = (panel["split"] == "test").to_numpy()
matched = panel["term"].isin(OWN_MAP).to_numpy()


def wape(mask, pred):
    return 100 * (weight[mask] * np.abs(rate[mask] - pred[mask])).sum() / (weight[mask] * rate[mask]).sum()


def fit_predict(search):
    X = panel[base_cols].copy()
    for name, arr in search.items():
        X[name] = arr
    model = XGBRegressor(
        n_estimators=300, max_depth=4, learning_rate=0.05, subsample=0.8, colsample_bytree=0.8,
        tree_method="hist", n_jobs=-1, random_state=0,
    )
    model.fit(X[tr], rate[tr], sample_weight=weight[tr])
    return np.clip(model.predict(X), 0, None)


variants = {"No search": {}}
for lag in LAGS:
    variants[f"VISUELLE trends, lag {lag}w"] = {f"vis_lag{lag}": vis_lag[lag]}
variants["VISUELLE trends, all lags 0-8w"] = {f"vis_lag{l}": vis_lag[l] for l in LAGS}
variants["VISUELLE + own keywords, all lags 0-8w"] = {
    **{f"vis_lag{l}": vis_lag[l] for l in LAGS}, **{f"own_lag{l}": own_lag[l] for l in LAGS}
}

e3_pred = {name: fit_predict(search) for name, search in variants.items()}

term_mean = {c: np.average(g["rate"], weights=g["n_in"]) for c, g in panel[tr].groupby("term_code")}
e3_pred["Training mean per term"] = panel["term_code"].map(term_mean).fillna(rate[tr].mean()).to_numpy()

e3_results = pd.DataFrame({
    name: {
        "val WAPE %": wape(va, p),
        "test WAPE %": wape(te, p),
        "test WAPE % (own-keyword tags only)": wape(te & matched, p),
    }
    for name, p in e3_pred.items()
}).T
display(e3_results.round(2))
```

```python
single = [f"VISUELLE trends, lag {l}w" for l in LAGS]
best_name = e3_results.loc[single, "val WAPE %"].idxmin()
print("Lag chosen on validation:", best_name)

# Paired bootstrap over test weeks: chosen lag vs no search.
test_dates, week_idx = np.unique(panel.loc[te, "date"].to_numpy(), return_inverse=True)
w_te, y_te = weight[te], rate[te]


def week_err(pred):
    return np.bincount(week_idx, weights=w_te * np.abs(y_te - pred[te]))


actual_w = np.bincount(week_idx, weights=w_te * y_te)
err_best, err_none = week_err(e3_pred[best_name]), week_err(e3_pred["No search"])
draws = np.random.default_rng(0).integers(0, len(test_dates), size=(1000, len(test_dates)))
diff = np.array([100 * (err_best[d].sum() - err_none[d].sum()) / actual_w[d].sum() for d in draws])
print(f"Test WAPE change vs no search: {diff.mean():+.2f} pp, 95% CI [{np.percentile(diff, 2.5):+.2f}, {np.percentile(diff, 97.5):+.2f}]")

# Does search lead sales? Correlation of within-term deviations: search at t - lag vs sales rate at t.
corr_rows = []
for lag in LAGS:
    d = pd.DataFrame({"term": panel["term"], "rate": rate, "s": vis_lag[lag]}).dropna()
    d[["rate", "s"]] -= d.groupby("term")[["rate", "s"]].transform("mean")
    corr_rows.append({"lag (weeks)": lag, "corr": d["rate"].corr(d["s"])})
lead_corr = pd.DataFrame(corr_rows).set_index("lag (weeks)")

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
lag_wape = e3_results.loc[single, ["val WAPE %", "test WAPE %"]].set_axis(list(LAGS))
lag_wape.plot(ax=axes[0], marker="o")
axes[0].axhline(e3_results.loc["No search", "test WAPE %"], color="grey", ls="--", label="test, no search")
axes[0].set(xlabel="search lag (weeks)", ylabel="WAPE %", title="Error by search lag")
axes[0].legend()
lead_corr["corr"].plot(ax=axes[1], marker="o", color="darkorange")
axes[1].set(xlabel="search lag (weeks)", ylabel="correlation", title="Search (t - lag) vs sales (t), within term")
plt.tight_layout()
plt.show()
```

### Forecasts at 4 and 8 weeks ahead 

The cells above use search at lags 0-8 to explain the *same* week's sales. For genuine forecasts, a prediction for week *t* at horizon *h* may only use what is known at the origin *t - h*: the weekly series up to *t - h*, search interest at lags of at least *h* weeks, and the planned release schedule and discounts. Models are fitted on the training seasons only; validation (SS19) and test (AW19) rows are written to `results/visuelle/predictions.csv`.

Baselines: `last_value` (series value at the origin) and `mean_last_4w` (average of the four weeks up to the origin, also repeated as `last4_avg` on every row). Models: `ridge` and `xgboost` (series value at the origin plus calendar and discount features), and `ridge_search` / `xgboost_search` (the same plus VISUELLE search at lags *h* to *h + 4*). `with_search` is yes for the `_search` models. `actual` is the mean units sold per in-stock listing of that product group in that week; `out_of_stock_week` is yes when any listing of the group was out of stock that week (those listing-weeks are excluded from `actual`) and `share_out_of_stock` is the share of listing-weeks affected. The data is weekly: every date is a Monday and consecutive dates are 7 days apart, so horizons 4 and 8 mean 4 and 8 weeks.

Scores (`results/visuelle/scores.csv`): MAE and RMSE in units per listing-week, WAPE % (weighted by in-stock listings), MASE (each item's absolute error scaled by that item's mean *h*-week naive error on the training seasons; below 1 beats last value) and `skill_vs_last_value` = 1 - MAE / MAE(last_value). MAE, RMSE and MASE are unweighted means over item-weeks.

```python
import os
from sklearn.impute import SimpleImputer
from sklearn.linear_model import Ridge
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

HORIZONS = [4, 8]
item = panel["attr"] + ":" + panel["term"]
known_cols = ["n_active", "age", "disc", "price", "week", "month"]  # release schedule and discounts are planned ahead
series = pd.DataFrame({"item": item, "split": panel["split"], "date": panel["date"], "rate": rate})
term_dummies = pd.get_dummies(item, dtype=float)
out_rows, scale_rows = [], []


def shifted(weeks_back):
    """Series value `weeks_back` weeks before each row's date, for the same item and split."""
    sh = series.assign(date=series["date"] + pd.Timedelta(weeks=weeks_back)).rename(columns={"rate": "r"})
    return series[["item", "split", "date"]].merge(sh, on=["item", "split", "date"], how="left")["r"].to_numpy()


for h in HORIZONS:
    origin_rate = shifted(h)
    search_h = pd.DataFrame({f"vis_lag{l}": lagged_search(vis, l) for l in range(h, h + 5)})

    usable = np.isfinite(origin_rate)  # same rows for every model, so the comparison is fair
    X_plain = pd.concat([panel[known_cols + ["attr_code", "term_code"]], pd.Series(origin_rate, name="origin_rate")], axis=1)
    X_search = pd.concat([X_plain, search_h], axis=1)
    X_ridge_plain = pd.concat([panel[known_cols], pd.Series(origin_rate, name="origin_rate"), term_dummies], axis=1)
    X_ridge_search = pd.concat([X_ridge_plain, search_h], axis=1)

    fit = usable & tr
    with np.errstate(all="ignore"):
        last4 = np.nanmean(np.vstack([shifted(h + s) for s in range(4)]), axis=0)
    preds = {"last_value": origin_rate, "mean_last_4w": last4}
    naive_scale = pd.Series(np.abs(rate - origin_rate)[fit]).groupby(item[fit].to_numpy()).mean()  # h-week naive error in training
    for name, X in [("ridge", X_ridge_plain), ("ridge_search", X_ridge_search)]:
        best = None
        for alpha in [1, 10, 100, 1000]:
            m = make_pipeline(SimpleImputer(strategy="median"), StandardScaler(), Ridge(alpha=alpha))
            m.fit(X[fit], rate[fit], ridge__sample_weight=weight[fit])
            p = np.clip(m.predict(X), 0, None)
            score = wape(usable & va, p)
            if best is None or score < best[0]:
                best = (score, p)
        preds[name] = best[1]
    for name, X in [("xgboost", X_plain), ("xgboost_search", X_search)]:
        m = XGBRegressor(n_estimators=300, max_depth=4, learning_rate=0.05, subsample=0.8, colsample_bytree=0.8,
                         tree_method="hist", n_jobs=-1, random_state=0)
        m.fit(X[fit], rate[fit], sample_weight=weight[fit])
        preds[name] = np.clip(m.predict(X), 0, None)

    for name, p in preds.items():
        for period, mask in [("validation", va), ("test", te)]:
            sel = usable & mask
            out_rows.append(pd.DataFrame({
                "date": panel.loc[sel, "date"].dt.strftime("%Y-%m-%d").to_numpy(), "item": item[sel].to_numpy(), "horizon": h,
                "actual": rate[sel], "predicted": p[sel], "model": name, "period": period, "last4_avg": last4[sel],
            }))
            scale_rows.append(pd.DataFrame({"item": item[sel].to_numpy(), "horizon": h, "scale": item[sel].map(naive_scale).to_numpy()}))

e3_predictions = pd.concat(out_rows, ignore_index=True)
e3_predictions["with_search"] = np.where(e3_predictions["model"].str.endswith("_search"), "yes", "no")

# Out-of-stock flag per item-week, from the corrected stock logic: listing-weeks with no stock at the start of the week.
flags = []
for attr in E_ATTRS:
    g = long.groupby([attr, "split", "date"]).agg(n_out=("in_stock", lambda x: int((~x).sum())), n_all=("in_stock", "size")).reset_index()
    g["item"] = attr + ":" + g[attr]
    flags.append(g[["item", "split", "date", "n_out", "n_all"]])
flags = pd.concat(flags)
flags["date"] = flags["date"].dt.strftime("%Y-%m-%d")
flags["period"] = flags["split"].map({"val": "validation", "test": "test"})
flags["out_of_stock_week"] = np.where(flags["n_out"] > 0, "yes", "no")
flags["share_out_of_stock"] = (flags["n_out"] / flags["n_all"]).round(4)
e3_predictions = e3_predictions.merge(
    flags[["item", "date", "period", "out_of_stock_week", "share_out_of_stock"]], on=["item", "date", "period"], how="left")
assert e3_predictions["out_of_stock_week"].notna().all()

e3_predictions = e3_predictions.sort_values(["period", "horizon", "model", "item", "date"])
os.makedirs("results/visuelle", exist_ok=True)
e3_predictions.to_csv("results/visuelle/predictions.csv", index=False)
print(f"Saved {len(e3_predictions)} rows to results/visuelle/predictions.csv")

# Score table. Weights are the number of in-stock listings behind each item-week.
w = panel[["date"]].assign(w=weight, item=item.to_numpy())
w["date"] = w["date"].dt.strftime("%Y-%m-%d")
sc = pd.concat(scale_rows).drop_duplicates(["item", "horizon"])
chk = (e3_predictions.merge(w.drop_duplicates(["item", "date"]), on=["item", "date"], how="left")
       .merge(sc, on=["item", "horizon"], how="left"))
chk["err"] = chk["actual"] - chk["predicted"]
chk["abs_err"] = chk["err"].abs()
chk["sq_err"] = chk["err"] ** 2
chk["scaled_err"] = chk["abs_err"] / chk["scale"]
chk["w_abs"] = chk["w"] * chk["abs_err"]
chk["w_act"] = chk["w"] * chk["actual"]


def score(g):
    return pd.Series({
        "n": len(g), "MAE": g["abs_err"].mean(), "RMSE": np.sqrt(g["sq_err"].mean()),
        "WAPE %": 100 * g["w_abs"].sum() / g["w_act"].sum(), "MASE": g["scaled_err"].mean(),
    })


score_table = chk.groupby(["period", "horizon", "model"]).apply(score).reset_index()
base_mae = score_table[score_table["model"] == "last_value"].set_index(["period", "horizon"])["MAE"]
score_table["skill_vs_last_value"] = 1 - score_table["MAE"] / score_table.set_index(["period", "horizon"]).index.map(base_mae).to_numpy()
score_table["n"] = score_table["n"].astype(int)
score_table.to_csv("results/visuelle/scores.csv", index=False)
display(score_table.round(3))

d = pd.to_datetime(pd.Series(sorted(e3_predictions["date"].unique())))
print("weekdays:", d.dt.day_name().unique().tolist(), "| gaps (days):", d.diff().dt.days.value_counts().to_dict(), "-> weekly data")
e3_predictions.head()
```
